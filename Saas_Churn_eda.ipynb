```python
# Total Customers & Churn Count
total_cust = len(df)
churned_cust = len(df[df['churn_label'] == 'Yes'])
churn_rate = (churned_cust / total_cust) * 100

print(f"Total Customers: {total_cust}")
print(f"Churned Customers: {churned_cust}")
print(f"Overall Churn Rate: {churn_rate:.2f}%")
```

```python
# FOR LOOP: Contract aur Payment Method par churn analysis
eda_cols = ['contract', 'payment_method']

for col in eda_cols:
    print(f"\n--- CHURN BY {col.upper()} ---")
    print(pd.crosstab(df[col], df['churn_label'], normalize='index') * 100)
```

```python

df['tenure_months'] = pd.to_numeric(df['tenure_months'], errors='coerce')
df['monthly_charges'] = pd.to_numeric(df['monthly_charges'], errors='coerce')

numeric_summary = df.groupby('churn_label')[['tenure_months', 'monthly_charges']].mean()
print("\n--- AVERAGE TENURE & CHARGES BY CHURN ---")
print(numeric_summary)
```

```python
# Dark Theme Styling
plt.style.use('dark_background')

# --- CHART 1: Churn by Contract Type (Bar Chart) ---
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='contract', hue='churn_label', palette=['#00E5FF', '#FF5252'])
plt.title("Churn Distribution across Contract Types", fontsize=12, fontweight='bold', color='white')
plt.xlabel("Contract Type")
plt.ylabel("Customer Count")
plt.tight_layout()
plt.savefig("churn_by_contract.png", dpi=300) 
plt.show()

# --- CHART 2: Tenure vs Monthly Charges (Scatter Plot) ---
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='tenure_months', y='monthly_charges', hue='churn_label', palette=['#00E5FF', '#FF5252'], alpha=0.7)
plt.title("Tenure vs Monthly Charges Analysis", fontsize=12, fontweight='bold', color='white')
plt.xlabel("Tenure (Months)")
plt.ylabel("Monthly Charges ($)")
plt.tight_layout()
plt.savefig("tenure_vs_charges.png", dpi=300) 
plt.show()
```

```python
# FOR LOOP: Customer Risk Category 
risk_list = []

for idx, row in df.iterrows():
    # Condition: Active customer + High bill (> $70) + Short tenure (<= 12 months)
    if row['churn_label'] == 'No' and row['monthly_charges'] > 70 and row['tenure_months'] <= 12:
        risk_list.append('High Risk Active')
    elif row['churn_label'] == 'Yes':
        risk_list.append('Churned')
    else:
        risk_list.append('Low Risk Active')


df['customer_risk_profile'] = risk_list

print("\n--- CUSTOMER RISK PROFILE BREAKDOWN ---")
print(df['customer_risk_profile'].value_counts())
```
